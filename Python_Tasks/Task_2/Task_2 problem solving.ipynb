```python
# Q1: Find the shortest word

# Observe: Input is a sentence; output is its shortest word.
# Understand: Find the word with the minimum length.
# Break: Split sentence, compare lengths, select the shortest.
# Logic: Keep updating the shortest word when a smaller one appears.

sentence = "Python makes coding easy"
words = sentence.split()

shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

# Code: Print the shortest word.
print(shortest)

# Test: Try another sentence.
test_sentence = "I love learning Python"
test_words = test_sentence.split()
test_shortest = test_words[0]

for word in test_words:
    if len(word) < len(test_shortest):
        test_shortest = word

print("Test output:", test_shortest)
```

```python
# Q2: Find the largest number

# Observe: Input is a list of numbers.
# Understand: Find the largest without using max().
# Break: Start with the first number and compare all numbers.
# Logic: Update largest whenever a bigger number is found.

numbers = [12, 45, 7, 89, 23]
largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)

# Test
test_numbers = [-5, -2, -10]
test_largest = test_numbers[0]

for number in test_numbers:
    if number > test_largest:
        test_largest = number

print("Test output:", test_largest)
```

```python
# Q6: Count vowels

# Observe: The string may contain uppercase and lowercase letters.
# Understand: Count all vowels.
# Break: Visit each character and check whether it is a vowel.
# Logic: Increase the count for every vowel.

text = "Education"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print(count)

# Test
test_text = "HELLO"
test_count = 0

for char in test_text:
    if char in vowels:
        test_count += 1

print("Test output:", test_count)
```

```python
# Q7: Count consonants

# Observe: The input can contain letters, spaces, digits and symbols.
# Understand: Count alphabetic characters that are not vowels.
# Break: Check each character for being a letter and a non-vowel.
# Logic: Increment the count only when both conditions are true.

text = "Python 123!"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char.isalpha() and char not in vowels:
        count += 1

print(count)

# Test
test_text = "Hello World!"
test_count = 0

for char in test_text:
    if char.isalpha() and char not in vowels:
        test_count += 1

print("Test output:", test_count)
```

```python
# Q8: Find the most frequent character

# Observe: Spaces must be ignored.
# Understand: Find the character with the highest frequency.
# Break: Count characters and compare their frequencies.
# Logic: Update only for a strictly higher frequency.

text = "banana".replace(" ", "")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

most_frequent = text[0]

for char in text:
    if frequency[char] > frequency[most_frequent]:
        most_frequent = char

print(most_frequent)

# Test
test_text = "aabbc"
test_text = test_text.replace(" ", "")
test_frequency = {}

for char in test_text:
    test_frequency[char] = test_frequency.get(char, 0) + 1

test_result = test_text[0]

for char in test_text:
    if test_frequency[char] > test_frequency[test_result]:
        test_result = char

print("Test output:", test_result)
```

```python
# Q10: Find repeated characters

# Observe: Each repeated character must appear only once in the result.
# Understand: Identify characters with frequency greater than one.
# Break: Count characters, then collect qualifying characters.
# Logic: Use a list to preserve order and avoid duplicates.

text = "programming".replace(" ", "")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

repeated = []

for char in text:
    if frequency[char] > 1 and char not in repeated:
        repeated.append(char)

print(repeated)

# Test
test_text = "hello"
test_frequency = {}

for char in test_text:
    test_frequency[char] = test_frequency.get(char, 0) + 1

test_repeated = []

for char in test_text:
    if test_frequency[char] > 1 and char not in test_repeated:
        test_repeated.append(char)

print("Test output:", test_repeated)
```

```python
# Q11: Words and their first characters

# Observe: Each word is a key and its first character is the value.
# Understand: Create a dictionary from a sentence.
# Break: Split the sentence and process each word.
# Logic: Map each word to word[0].

sentence = "Python Data Engineering"
words = sentence.split()
result = {}

for word in words:
    result[word] = word[0]

print(result)

# Test
test_sentence = "Apple Mango"
test_result = {}

for word in test_sentence.split():
    test_result[word] = word[0]

print("Test output:", test_result)
```

```python
# Q14: Numbers and their squares

# Observe: Each number is the key and its square is the value.
# Understand: Create a dictionary from a list of numbers.
# Break: Visit every number and calculate number * number.
# Logic: Store each number and its square.

numbers = [2, 3, 4, 5]
result = {}

for number in numbers:
    result[number] = number ** 2

print(result)

# Test
test_numbers = [6, 7]
test_result = {}

for number in test_numbers:
    test_result[number] = number ** 2

print("Test output:", test_result)
```

```python
# Q15: Numbers and their even/odd status

# Observe: Each number must be labelled Even or Odd.
# Understand: Determine whether each number is divisible by 2.
# Break: Visit each number and check its remainder.
# Logic: Remainder 0 means Even; otherwise, Odd.

numbers = [1, 2, 3, 4, 5]
result = {}

for number in numbers:
    if number % 2 == 0:
        result[number] = "Even"
    else:
        result[number] = "Odd"

print(result)

# Test
test_numbers = [10, 11]
test_result = {}

for number in test_numbers:
    if number % 2 == 0:
        test_result[number] = "Even"
    else:
        test_result[number] = "Odd"

print("Test output:", test_result)
```

```python
# Q16: Find the most frequent word

# Observe: Uppercase and lowercase versions count as the same word.
# Understand: Find the word with the highest frequency.
# Break: Convert words to lowercase and count occurrences.
# Logic: Keep the first word having the maximum frequency.

sentence = "Python is easy and Python is powerful"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

most_frequent = words[0]

for word in words:
    if frequency[word] > frequency[most_frequent]:
        most_frequent = word

print(most_frequent)

# Test
test_sentence = "apple mango APPLE orange"
test_words = test_sentence.lower().split()
test_frequency = {}

for word in test_words:
    test_frequency[word] = test_frequency.get(word, 0) + 1

test_result = test_words[0]

for word in test_words:
    if test_frequency[word] > test_frequency[test_result]:
        test_result = word

print("Test output:", test_result)
```

```python
# Q18: Find the first unique word

# Observe: A word is unique if it appears exactly once.
# Understand: Return the first unique word in the sentence.
# Break: Count lowercase words and examine them in order.
# Logic: Print the first word whose frequency is one.

sentence = "apple banana apple mango banana orange"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

result = None

for word in words:
    if frequency[word] == 1:
        result = word
        break

print(result)

# Test
test_sentence = "red blue red green"
test_words = test_sentence.lower().split()
test_frequency = {}

for word in test_words:
    test_frequency[word] = test_frequency.get(word, 0) + 1

test_result = None

for word in test_words:
    if test_frequency[word] == 1:
        test_result = word
        break

print("Test output:", test_result)
```
