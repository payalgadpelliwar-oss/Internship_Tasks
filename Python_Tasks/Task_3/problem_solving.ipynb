Q1
1. Start with a sentence.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Count each word and store its frequency in the dictionary.
5. Return the dictionary.

```python
def word_frequency(sentence):
    result = {}

    for word in sentence.split():
        result[word] = result.get(word, 0) + 1

    return result

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("hello hello world"))
```

Q2
1. Start with a word.
2. Create an empty dictionary.
3. Visit each character in the word.
4. Count each character using the dictionary.
5. Return the dictionary.

```python
def character_frequency(word):
    result = {}

    for char in word:
        result[char] = result.get(char, 0) + 1

    return result

print(character_frequency("banana"))
print(character_frequency("hello"))
```

Q3
1. Start with a word.
2. Count the frequency of every character.
3. Visit the characters in their original order.
4. Return the first character whose frequency is one.
5. If no such character exists, return None.

```python
def first_non_repeating(word):
    frequency = {}

    for char in word:
        frequency[char] = frequency.get(char, 0) + 1

    for char in word:
        if frequency[char] == 1:
            return char

    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))
```

Q4
1. Take two strings.
2. Count the frequency of each character in both strings.
3. Compare the two dictionaries.
4. Return True if they are equal; otherwise, return False.

```python
def are_anagrams(s1, s2):
    return character_frequency(s1) == character_frequency(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
```

Q5
1. Start with a string.
2. Create an empty set.
3. Visit each character from left to right.
4. If the character is already in the set, return it.
5. Otherwise, add it to the set.
6. If no character repeats, return None.


```python
def first_repeating_char(text):
    seen = set()

    for char in text:
        if char in seen:
            return char
        seen.add(char)

    return None

print(first_repeating_char("programming"))
print(first_repeating_char("abc"))
```
